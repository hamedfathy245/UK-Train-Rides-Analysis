```python
price_Q1 = df["Price"].quantile(0.25)
price_Q3 = df["Price"].quantile(0.75)

price_IQR = price_Q3 - price_Q1

price_LWRB = price_Q1 - 1.5 * price_IQR
price_UPPB = price_Q3 + 1.5 * price_IQR

print("Lower Bound:", price_LWRB)
print("Upper Bound:", price_UPPB)
```

```python
price_outlier_count = (
    (df["Price"] < price_LWRB) |
    (df["Price"] > price_UPPB)
).sum()

price_outlier_percentage = (
    price_outlier_count / len(df)
) * 100

print("Number of Price Outliers:", price_outlier_count)
print(f"Price Outliers Percentage: {price_outlier_percentage:.2f}%")
```

```python
df[
    (df["Price"] < price_LWRB) |
    (df["Price"] > price_UPPB)
][
    ["Price", "Ticket Type", "Ticket Class", "Purchase Type"]
].head(10)
```

```python
for (ticket_type, ticket_class), group in df.groupby(
    ["Ticket Type", "Ticket Class"] ):

    price_Q1 = group["Price"].quantile(0.25)
    price_Q3 = group["Price"].quantile(0.75)

    price_IQR = price_Q3 - price_Q1

    price_UPPB = price_Q3 + 1.5 * price_IQR
    price_LWRB = price_Q1 - 1.5 * price_IQR

    outlier_count = (
        (group["Price"] < price_LWRB) |
        (group["Price"] > price_UPPB)
    ).sum()

    outlier_percentage = (outlier_count / len(group)) * 100

    print(ticket_type, "-", ticket_class)
    print("Lower Bound:", price_LWRB)
    print("Upper Bound:", price_UPPB)
    print("Outliers:", outlier_count)
    print(f"Outliers Percentage: {outlier_percentage:.2f}%")
    print("=" * 40)
```
